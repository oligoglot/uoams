Animal rehab data exploration and visualisation

```python
import pandas as pd
import numpy as np
data = pd.read_csv("Urban_Park_Ranger_Animal_Condition_Response.csv")
print(data.shape)
data.head(20)
```

Preprocess data

```python
reldf = data[['Date and Time of initial call', 'Date and time of Ranger response', 'Species Status', 'Animal Class', 'Age', 'Animal Condition', '# of Animals', 'Call Source', 'Final Ranger Action', 'Hours spent monitoring']]
reldf.columns = ['Time of call', 'Time of response', 'Species Type', 'Animal Class', 'Age', 'Animal Condition', 'Count of animals', 'Call Source', 'Final Ranger Action', 'Monitoring hours']
reldf['Time of call'] = pd.to_datetime(reldf['Time of call'])
reldf['Time of response'] = pd.to_datetime(reldf['Time of response'])
reldf['Event year'] = reldf['Time of response'].dt.year
reldf['Time to react'] = reldf['Time of response'] - reldf['Time of call']
reldf['Time to react'] = reldf['Time to react'].dt.seconds/60
reldf['Time to react'] = pd.to_numeric(reldf['Time to react'], errors='coerce').fillna(0).astype(np.int32)
animaltypes = {'Birds' : 'Birds',
                'Coyotes' : 'Mammals',
                'Deer' : 'Mammals',
                'Domestic' : 'Domestic',
                'Domestic;#Birds' : 'Domestic',
                'Domestic;#Small Mammals-non RVS' : 'Domestic',
                'Fish-numerous quantity' : 'Fish',
                'Marine Mammals-seals only' : 'Mammals',
                'Raptors' : 'Birds',
                'Small Mammals-RVS' : 'Mammals',
                'Small Mammals-non RVS' : 'Mammals',
                'Terrestrial Reptile or Amphibian' : 'Herps',
                'Birds;#Domestic' : 'Birds',
                'Fish-numerous quantity;#Terrestrial Reptile or Amphibian' : 'Fish',
                'Marine Mammals-whales, Dolphin' : 'Mammals',
                'Marine Reptiles' : 'Herps',
                'Small Mammals-RVS;#Domestic' : 'Mammals',
                'Non Native Fish-(invasive)' : 'Fish',
                'Rare, Endangered, Dangerous' : 'Rare'}
reldf['Animal Class'] = reldf['Animal Class'].map(animaltypes)
reldf.to_csv("outputtable.csv", index=False)
reldf.head(100)
```

1) How has the number of animals sent to rehabilitators or animal care centres changed over time?

```python
filtered_df = reldf[reldf['Final Ranger Action'].isin(['ACC', 'Rehabilitator'])]
animalcountchange = filtered_df.groupby(['Species Type', 'Event year', 'Final Ranger Action']).agg({'Count of animals': 'sum', 'Time to react': 'mean'}).reset_index()
animalcountchange.to_csv("animalcountchange.csv", index=False)
```

2) How has the type of animal being reported changed over time?

```python
animaltypechange = reldf.groupby(['Species Type', 'Event year']).agg({'Count of animals': 'sum'}).reset_index()
animaltypechange.to_csv("animaltypechange.csv", index=False)
```

3) How has the source of information (call source) changed over time?

```python
callsourcechange = reldf.groupby(['Species Type', 'Event year', 'Call Source']).agg({'Count of animals':'count'}).reset_index()
callsourcechange.to_csv("callsourcechange.csv")
```
