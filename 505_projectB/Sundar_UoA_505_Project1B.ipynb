```python
from bs4 import BeautifulSoup
import unittest

soup = BeautifulSoup(html, 'html.parser')
print(soup.title.string)

def fetch_and_parse(url):
    needed_headers = {'User-Agent': "Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/92.0.4515.131 Safari/537.36"}
    try:
        response = requests.get(url, headers = needed_headers)
        response.raise_for_status()
    except requests.exceptions.HTTPError as http_err:
        raise Exception(http_err)
    except requests.exceptions.RequestException as req_exc:
        raise Exception(req_exc)
    return BeautifulSoup(response.text, 'html.parser')

class Tester(unittest.TestCase):
    def test_valid(self):
        soup = fetch_and_parse("https://www.arizona.edu")
        # print(soup.title.string)
        assert len(soup.text) > 0
        
    def test_invalid(self):
        with self.assertRaises(Exception) as context:
            fetch_and_parse("https://www.arizona.edu/nonexistentpage")

tester = Tester()        
tester.test_valid()
tester.test_invalid()
```

B ) Note that some movies might not have a user rating. Your function should be able to parse

be appended to the list created in Q4b

```python
#4a
def get_all_titles(soup):
    titles = []
    movies = soup.find_all('div', 'card style_1')
    for movie in movies:
        title = movie.find('h2')
        titles.append(title.text)
    return titles

print(get_all_titles(soup))
```

```python
#4b
def get_all_ratings(soup):
    ratings = []
    movies = soup.find_all('div', 'card style_1')
    for movie in movies:
        score = movie.find(attrs={'data-percent': True})
        if score is None:
            ratings.append("not rated")
        else:
            ratings.append(score.get('data-percent'))
    return ratings

print(get_all_ratings(soup))
```

```python
#4c
def get_all_movies(soup):
    movies = []
    for movie in soup.find_all('div', 'card style_1'):
        movies.append(movie.contents)
    return movies

movies = get_all_movies(soup)
if len(movies) > 0:
    print(movies[0])
```
