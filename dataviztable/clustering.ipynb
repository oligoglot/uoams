# Week 5 - Clustering

### Loading libraries and creating the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler

# Setting seed for reproducibility
np.random.seed(0)

# Create a mock-up dataset
data = {
    'Age': np.random.randint(20, 70, 100),
    'Income': np.random.randint(30000, 150000, 100),
    'Score': np.random.randint(1, 100, 100)
}

df = pd.DataFrame(data)

# Standardize the data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
df
```

### Next, let's use K-Means for hard clustering:

```python
# Define the model
kmeans = KMeans(n_clusters=3, random_state=0)

# Fit and predict
kmeans_clusters = kmeans.fit_predict(df_scaled)

# Plot the clusters
plt.scatter(df_scaled[kmeans_clusters == 0, 0], df_scaled[kmeans_clusters == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(df_scaled[kmeans_clusters == 1, 0], df_scaled[kmeans_clusters == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(df_scaled[kmeans_clusters == 2, 0], df_scaled[kmeans_clusters == 2, 1], s = 100, c = 'green', label = 'Cluster 3')

# Plot the centroids
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 300, c = 'black', label = 'Centroids')

plt.legend()
plt.show()
```

### Now, let's use Gaussian Mixture Model for soft clustering:

```python
# Define the model
gmm = GaussianMixture(n_components=3, random_state=0)

# Fit and predict
gmm_clusters = gmm.fit_predict(df_scaled)

# Plot the clusters
plt.scatter(df_scaled[gmm_clusters == 0, 0], df_scaled[gmm_clusters == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(df_scaled[gmm_clusters == 1, 0], df_scaled[gmm_clusters == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(df_scaled[gmm_clusters == 2, 0], df_scaled[gmm_clusters == 2, 1], s = 100, c = 'green', label = 'Cluster 3')

plt.legend()
plt.show()
```
