## Question 2

Assume that you collected a dataset of 100 observations containing a single predictor and a quantitative response. Then, you decided to fit a

(i) linear regression model to the data, as well as a separate

(ii) cubic regression, i.e. $Y = β_{0} + β_{1}X^{1} + β_{2}X^{2} + β_{3}X^{3} + ϵ$

### a) Suppose that the true relationship between $X$ and $Y$ is linear, i.e. $Y = β_{0} + β_{1}X + ϵ$. Consider the training residual sum of squares (_RSS_) for the linear regression, and also the training _RSS_ for the cubic regression. Would we expect one to be lower than the other, would we expect them to be the same, or is there not enough information to tell? Justify your answer

The cubic regression model risks overfitting unless the co-efficient for the higher order co-efficients (β_{2}, β_{3}) approaches zero. However, the RSS on training data will be lower for an overfit model (the cubic regression model in our case), because more datapoints can be fit closer to the resulting curve. Overfitting can be minimal if there is a large number of training data points which follow a more natural distribution. In the best case scenario β_{2} and β_{3} become 0, and, in that case, the cubic model will perform as well as the linear model.

### b) Answer part (a) using the test _RSS_ rather than training _RSS_.

If the test data follows a natural distribution, an overfit model will err more often. Given this, a cubic model can potentially err more often than the linear model. A cubic model is likely to have a higher RSS against test data with the same caveat as above (unless β_{2} and β_{3} become 0).

### c) Suppose that the true relationship between $X$ and $Y$ is not linear. However, we don’t know how far it is from linear. Consider the training _RSS_ for the linear regression, and also the training _RSS_ for the cubic regression. Would we expect one to be lower than the other, would we expect them to be the same, or is there not enough information to tell? Justify your answer.

If it is supra-linear, the cubic model can overfit at worst, but that'd only make it err less on training data. Hence training RSS would be lower for a cubic model.

### d) Answer (c) using test _RSS_ rather than training _RSS._

With test data, however, it depends on how far it is from linear. If it is more than quadratic, a cubic model might perform better than linear and hence might have a lower RSS on test data. If it's below quadratic and closer to a linear model, the linear model will perform better on a properly distributed test data and hence have a lower RSS.

### a) Discuss the implications that uncertainty in $X$ would have on parameter estimates based on linear regression models. Should we care about error in $X$ when fitting OLS regressions?



Most real world data will have some uncertainty. It can be due to human error, measuring instrument error or because of a missing feature. This could potentially adversely affect parameter estimates if the error datapoints are significant in number or magnitude. A data scientist should care about such errors and mitigate as much as possible. Certain data points can be eliminated by applying domain knowledge (temperatures way outside of known ranges for a season or geography should be filteres, for example), outlier detection, smoothing etc. Uncertainty in X can cause bias if not addressed.

### b) Provide a brief discussion on the concept of regression dilution.




Due to excessive noise in the independent variable, there could be a bias forcing the slope to tend towards 0. This essentially dilutes the prediction of the regression because the slope is impacted by error rather than the variable itself. This is called regression dilution.

### c) Are there any linear regression models that account for uncertainty in $X$ and $Y$? Please do some independent research and cite your sources.


Correlation disattenuation can be employed to account for any bias that lingers after regular noise elimination and smoothing techniques.

### d) Distinguish between least squares and maximum likelihood in the context of linear regression models



They are essentially same when the distribution is normal. When it's not a normal distribution, they can be mapped using a link function.
